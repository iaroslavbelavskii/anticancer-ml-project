## Задание 3: Функции

**Описание:** Создайте функции для работы с данными пациентов.

**Требования:**
- Создайте функцию `calculate_bmi(height, weight)` - индекс массы тела
- Создайте функцию `count_positive_cases(diagnoses)` - количество положительных диагнозов
- Создайте функцию `patient_info(name, age, diagnosis)` - выведите информацию о пациенте
- Используйте значения по умолчанию в одной из функций

```python
def calculate_bmi(height, weight):
    bmi = weight / height**2
    return bmi

def count_positive_cases(diagnoses):
    count = sum(diagnoses)
    return count

def patient_info(name, age, has_cancer):
    info = f"Patient Name: {name}, Age: {age}, Has Cancer: {has_cancer}"
    return info
```

## Задание 4: Основы Pandas

**Описание:** Создайте DataFrame с данными пациентов.

**Требования:**
- Импортируйте pandas
- Создайте DataFrame с колонками: имя, возраст, диагноз, возможный исход
- Выведите первые 3 строки
- Выведите информацию о DataFrame (info())
- Вычислите статистику (describe())

```python
import pandas as pd

names = ["Иван Петров", "Мария Сидорова", "Петр Иванов", "Анна Смирнова", "Сергей Кузнецов", "Елена Волкова"]
age = [45, 30, 60, 25, 50, 60]
diagnoses = [True, False, True, False, True, False]
otcomes = ["Good", "Poor", "Good", "Excellent", "Poor", "Good"]

df = pd.DataFrame({
    "Name": names,
    "Age": age,
    "Has Cancer": diagnoses,
    "Possible Outcome": otcomes
})

print(df[1:4])
print(df.info())
print(df.describe())
```
